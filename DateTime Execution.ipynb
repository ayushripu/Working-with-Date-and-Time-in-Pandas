# Convert String to DateTime

```python
df['date'] = pd.to_datetime(df['date'], format='%d/%m/%Y')
```

```python
df.info()
```

### Now Pandas understands:
`This is a real date.`

```python
df.head()
```

```python
df1 = df['date'].dt.day
df1.head()
```

```python
df2 = df['date'].dt.month
df2.head()
```

```python
df3 = df['date'].dt.year
df3.head()
```

```python
df4 = df['date'].max()
df4
```

```python
df5 = df['date'].min()
df5
```

```python
df6 = df.loc[df['date'] == '2024-03-18']
df6
```

```python
df9 = df.loc[df['date'].between('2024-03-25', '2024-05-13')]
df9
```

### Difference in Two Datetime Columns

```python
df23 = pd.read_csv(r'C:\Users\ayush\OneDrive\Desktop\ExcelData\fast_delivery.csv')

df23.head()
```

```python
df23.info()
```

```python
df23['promised_delivery_time'] = pd.to_datetime(df23['promised_delivery_time'], format = 'mixed')

df23.head()
```

```python
df23['actual_delivery_time'] = pd.to_datetime(df23['actual_delivery_time'], format = 'mixed')

df23.head()
```

```python
df23.info()
```

```python
df24= df23.copy()

df24['Difference'] = df23['promised_delivery_time'] - df23['actual_delivery_time']

df24
```

```python
df25= df23.copy()

df25['Difference'] = (df23['promised_delivery_time'] - df23['actual_delivery_time']).dt.total_seconds()

df25
```

```python
df26= df24.copy()

df26['Difference'] = df24['Difference'].dt.total_seconds()

df26
```

```python
df27= df24.copy()

df27['Difference'] = df26['Difference']/60

df27
```
